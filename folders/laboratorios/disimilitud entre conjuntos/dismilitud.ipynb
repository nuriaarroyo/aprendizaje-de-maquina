### Entrada

**Input:**  
Matriz de distancias $D_{n \times n}$  
Donde $D = [d_{ij}]$, y $d_{ij}$ representa la disimilitud entre los puntos $i$ y $j$.

---



### Función: Distancia de un punto a un conjunto (promedio)

```plaintext
Función conjunto_al_punto(C_set, a, D):
    C_temp ← C
    sum ← 0
    c ← 0
    while C_temp ≠ ∅:
        x ← x ∈ C_temp
        C_temp ← C_temp \ {x}
        sum ← sum + d_{ax}
        c ← c + 1
    distancia ← sum / c
    retornar distancia

```
---

### Función medoide
```plaintext
Función medoide(C, D):
    C_temp ← C
    min ← ∞
    medoide ← None
    while C_temp ≠ ∅:
        x ← x ∈ C_temp
        dist ← conjunto_al_punto(C, x, D)
        if dist < min:
            min ← dist
            medoide ← x
        C_temp ← C_temp \ {x}
    retornar medoide
```
---

### Función: Distancia entre dos conjuntos (medoide a medoide)

```plaintext    
Función medoide_a_medoide(C_set, E_set, D):
    c ← medoide(C)
    e ← medoide(E)
    distancia ← d_{ce}
    retornar distancia
```
---

## Algoritmo principal 1

### Distancia de un punto a un conjunto

**Input:**

- Conjunto de puntos $ C $
- Punto  $x$
- Matriz de distancia $ D_{n \times n} $

**Paso:**

```plaintext
distancia = conjunto_al_punto(C, x, D)
```
---

## Algoritmo principal 2

### Distancia de un conjunto a otro conjunto (medoide a medoide)

**Input:**

- Conjunto de puntos $ C $
- Conjunto de puntos $ E $
- Matriz de distancia $ D_{n \times n} $

**Paso:**

```plaintext
distancia = medoide_a_medoide(C, E, D)
```
